```python
features = [
    "amount_inr",
    "payment_method",
    "failure_type",
    "hour_of_day",
    "minutes_since_failure",
    "retry_count",
    "previous_payments",
    "previous_successes",
    "previous_success_rate"
]
```

```python
target = "recovered"
```

```python
X=df[features]
Y=df[target]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "payment_method",
    "failure_type"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=5000))
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
y_prob = model.predict_proba(X_test)

print(y_prob[:100])
```

```python
recovery_probability = y_prob[:, 1]

print(recovery_probability[:10])
```

```python
results = X_test.copy()

results["actual_recovered"] = y_test.values
results["recovery_probability"] = recovery_probability
results["predicted_recovered"] = y_pred

results.head(10)
```

```python
print("Baseline Model Performance")
print("--------------------------")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2%}")

print("\nRecovery prediction performance:")
print(classification_report(y_test, y_pred))
```

```python
threshold = 0.40

y_pred_40 = (recovery_probability >= threshold).astype(int)

print(classification_report(y_test, y_pred_40))
```

# Day 2 — ML Baseline Complete

## Objective

Build a baseline machine learning model that estimates whether a
failed payment is likely to be recovered based on historical
payment behaviour.

## Dataset

- 10,000 historical payment records
- 12 features
- Failure types:
  - card_declined
  - insufficient_funds
  - network_timeout
  - checkout_abandoned

## Baseline Model

Logistic Regression with:
- One-hot encoding for categorical features
- Numerical features passed through directly
- 80/20 train-test split

## Results

Baseline threshold (50%):
- Accuracy: 71.05%
- Recovery precision: 64%
- Recovery recall: 44%
- Recovery F1-score: 52%

Experimental threshold (40%):
- Recovery precision: 57%
- Recovery recall: 62%
- Recovery F1-score: 60%

## Conclusion

The historical payment data contains useful predictive signals for
payment recovery.

The model outputs a recovery probability rather than directly
taking an action. This probability will be provided to the RevivePay
decision engine, which will combine it with business and safety
rules before deciding whether to retry, wait, escalate, or stop.

The Logistic Regression model is currently a baseline and can be
improved later if necessary.
